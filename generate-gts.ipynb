```python
import numpy as np

file_path = './glove.42B.300d.txt' 

embeddings = []
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        values = line.split()[1:]
        embeddings.append(np.array(values, dtype=np.float32))

dataset = np.array(embeddings)

print(dataset.shape)
```

```python
import struct

def write_fvecs(filename, tensor_data, tp='<f'):
    # Open the file in binary write mode
    with open(filename, 'wb') as f:
        # Iterate over each vector in the tensor
        for vector in tensor_data:
            # Get the dimensionality (d)
            d = len(vector)

            # Write the dimension as a 4-byte integer (little-endian)
            f.write(struct.pack('<I', d))  # '<I' is for unsigned int (4 bytes, little-endian)

            # Write the actual vector elements as 4-byte floats (little-endian)
            for value in vector:
                f.write(struct.pack(tp, value.item()))  # '<f' is for float (4 bytes, little-endian)
```

```python
write_fvecs('db.fvecs', dataset)
```

```python
num_queries = 10000

sampled_indices = np.random.randint(0, dataset.shape[0], size=num_queries)
sampled_vectors = dataset[sampled_indices]
write_fvecs('queries.fvecs', sampled_vectors)
```

```python
import faiss

dimension = sampled_vectors.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(dataset)

K = 100
distances, indices = index.search(sampled_vectors, K)
print(indices[0])
```

```python
write_fvecs('distances-100.fvecs', distances)
write_fvecs('indices-100.fvecs', indices, "<i")
```
